### **Figure 3. Pyrethroid target-site resistance**

- Vgsc gene target-site resistance - AGAP004707-RD

```python
df_samples_gambia.pivot_table(
    index=["admin1_name", "admin2_name", "year"],
    columns="taxon",
    values="sample_id",
    aggfunc="count",
    fill_value=0,
)
```

```python
df_vgsc_freq_gambia = ag3.aa_allele_frequencies(
    transcript="AGAP004707-RD",
    sample_query=gambia_sample_query,
    cohorts="admin2_year",
)
df_vgsc_freq_gambia
```

```python
ag3.plot_frequencies_heatmap(
    df_vgsc_freq_gambia.query("max_af > 0.05")
)
```

```python
df_vgsc_freq_gambia_admin2_year = (
    df_samples_gambia
    .groupby("cohort_admin2_year")
    .aggregate({
        "latitude": "mean",
        "longitude": "mean",
        "sample_id": "count",
    })
    .query("sample_id >= 10")
    .sort_values("longitude")
)
df_vgsc_freq_gambia_admin2_year
```

```python
all_cohorts_admin2_year = df_vgsc_freq_gambia_admin2_year.index.tolist()
all_cohorts_admin2_year
```

```python
ag3.plot_frequencies_heatmap(
    df_vgsc_freq_gambia.query("max_af > 0.05")[["label"] + [f"frq_{c}" for c in all_cohorts_admin2_year]]
)
```

```python
frequency_columns = [
    col for col in df_vgsc_freq_gambia.columns
    if col.startswith("frq_")
]
frequency_columns
```

```python
vgsc_aa_effect_df = df_vgsc_freq_gambia[["effect", "impact"]]
vgsc_aa_effect_df
```

```python
arab_cohorts_admin2_year_gambia = (
    df_samples_gambia
    .query("taxon == 'arabiensis'")
    .groupby("cohort_admin2_year")
    .aggregate({
        "latitude": "mean",
        "longitude": "mean",
        "sample_id": "count",
    })
    .query("sample_id >= 10")
    .sort_values("longitude")
)
arab_cohorts_admin2_year_gambia
```

```python
arab_cohorts_admin2_year = arab_cohorts_admin2_year_gambia.index.tolist()
arab_cohorts_admin2_year
```

```python
ag3.plot_frequencies_heatmap(
    df_vgsc_freq_gambia.query("max_af > 0.05")[["label"] + [f"frq_{c}" for c in arab_cohorts_admin2_year]],
    row_height=30,
    color_continuous_scale='Turbo'
)
```

```python
df_vgsc_freq_gambia.head(3)
```

```python
# Select the knowns amino-acid changes associated with vsgc target site IR in addition to the 4 with max_af > 0.05
selected_aa_changes = ["T664S", "H721Q", "T791M", "L995S", "L995F", "A1553T", "N1570Y", "A1746S", "P1874L"]
```

```python
df_vgsc_freq_gambia = ag3.aa_allele_frequencies(
    transcript="AGAP004707-RD",
    sample_query=gambia_sample_query,
    cohorts="admin2_year",
    drop_invariant=False,
)
```

```python
#Let's plot SNPs frequences for the selected aa of interest to see
df_vgsc_freq_gambia_selected = (
    df_vgsc_freq_gambia
    .loc[
        df_vgsc_freq_gambia.index.get_level_values("aa_change")
        .isin(selected_aa_changes)
    ]
)

# Plot the heatmap
ag3.plot_frequencies_heatmap(
    df_vgsc_freq_gambia_selected[
        ["label"] + [f"frq_{c}" for c in arab_cohorts_admin2_year]
    ],
    row_height=30,
    color_continuous_scale="Turbo"
)
```
