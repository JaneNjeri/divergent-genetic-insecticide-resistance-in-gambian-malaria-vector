### **Cyp450_keap_gste locus CNV Freq -- Supplementary Figure S4a **

```python
################ Cyt P450 // Gste // Keap 1 metabolik CNV Freq ###########
cyp6aap_region = "2R:28,480,000-28,510,000"
cyp9k1_region = "X:15,240,000-15,250,000"
#cyp6mz_region = "3R:6,924,000-6,980,000" # This is displaying 3 more genes i dont wont so let reduce the range of 6mz genes !!!
cyp6mz_region = "3R:6,970,000-6,980,000"  # THe first gene of interest start with at 6971582 and the 2nd end at 6975563
keap1_region = "2R:40,926,195-40,945,169"
gste1_region = "3R:28,598,871-28,599,816"
gste2_region = "3R:28,597,652-28,598,640"
cyp_regions = [keap1_region, cyp6aap_region, cyp6mz_region, cyp9k1_region, gste1_region, gste2_region]
```

```python
df_cyp_keap_gste_cnv_freq = ag3.gene_cnv_frequencies(
    sample_query="country == 'Gambia, The' and taxon == 'arabiensis' and sample_id!='VBS47740-6366STDY9854349'and release == '3.5'",
    cohorts="admin2_year",
    region=cyp_regions,
    sample_sets="3.5",  # needed because CNV data not released for 3.1
   # drop_invariant=False,
)
df_cyp_keap_gste_cnv_freq
```

```python
# define and select frequency_columns variable
frequency_columns = [
    col for col in df_cyp_keap_gste_cnv_freq.columns
    if col.startswith("frq_")
]
frequency_columns
```

```python
# To display the DataFrame selecting the defined frequency_columns + Max_af and windows columns
#gene_id , gene_name and cnv_type are index
df_cyp_keap_gste_cnv_freq[frequency_columns + ["max_af", "windows"]]
```

```python
df_cyp_keap_gste_cnv_freq.columns
```

```python
arab_cnv_cohorts_admin2_year = [ #to plot it following the geographical alignement of the cohorts
    "GM-W_Jeshwang_arab_2019",
    "GM-W_Kombo-Central_arab_2019",
    "GM-M_Lower-Fuladu-West_arab_2019",
    "GM-U_Jimara_arab_2019",
    "GM-U_Basse_arab_2019"
]
```

```python
ag3.plot_frequencies_heatmap(
    df_cyp_keap_gste_cnv_freq .query("max_af >= 0.00")[["label"] + [f"frq_{c}" for c in arab_cnv_cohorts_admin2_year]],
    height=600, width=800,
    title="Gene CNV frequencies, Keap1(AGAP003645)+ Cyp6aa/p + Cyp9k1 + Cym6mz + gst1/2"
)
```

```python
ag3.plot_frequencies_heatmap(
    df_cyp_keap_gste_cnv_freq.query("max_af >= 0.00")[["label"] + [f"frq_{c}" for c in arab_cnv_cohorts_admin2_year]],
    height=600, width=700, color_continuous_scale = "turbo",
    title="Gene CNV frequencies, Keap1(AGAP003645)+ Cyp6aa/p + Cyp9k1 + Cym6mz + gst1/2",
)
```

```python
##################Supplementary Figure S4a ##################

# Rename the specific label in the DataFrame
df_cyp_keap_gste_cnv_freq['label'] = df_cyp_keap_gste_cnv_freq['label'].replace(
    'AGAP003645 del', 'AGAP003645 (Keap1) del'
)
# Lets increase the size of the legends for better displaying !!!
fig = ag3.plot_frequencies_heatmap (
   df_cyp_keap_gste_cnv_freq.query("max_af >= 0.00")[["label"] + [f"frq_{c}" for c in arab_cnv_cohorts_admin2_year]],
    height=700,
    width=850,
    color_continuous_scale="turbo", #["lightblue", "purple"]
    title="Gene CNV frequencies, Keap1 + Cyp6aa/p + Cyp9k1 + Cym6mz + gste1/2",
    show=False, # Set show=False to return the figure object for modification
)

# Increase colorbar legend font size (title and tick labels) and text inside cells font size
fig.update_layout(
    legend=dict(
       font=dict(size=16),
    ),
    coloraxis_colorbar=dict(
        title_font=dict(size=16), # Adjust colorbar title font size
        tickfont=dict(size=16)   # Adjust colorbar tick label font size
    ),
    xaxis=dict( # Update x-axis properties for cohorts
        tickfont=dict(size=16), # Adjust font size for x-axis tick labels (cohorts) - Increased size
        title_font=dict(size=16) # Adjust font size for x-axis title
    ),
    yaxis=dict( # Update y-axis properties for variants
        tickfont=dict(size=16), # Adjust font size for y-axis tick labels (variants) - Increased size
        title_font=dict(size=16) # Adjust font size for y-axis title
    )
)

# Set the font size for the text inside the heatmap cells
fig.data[0].textfont = dict(size=16
                            )

# Display the modified figure
fig.show()
```
