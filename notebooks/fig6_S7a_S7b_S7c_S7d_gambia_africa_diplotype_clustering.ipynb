```python
africa_sample_sets = ['AG1000G-BF-B','AG1000G-CM-C','1272-VO-GM-OPONDO-VMF00160','AG1000G-ML-B','1314-VO-BF-KIENTEGA-KIMA-BF-2104']
```

### **Figure 6. diplotype clustering at Coeae2-7g locus in An. arabiensis from the Gambia & Tanzania**

```python
coeae2_7g_diplo = ag3.plot_diplotype_clustering_advanced(
    region='2L:37,282,152-37,298,223',
    snp_transcript='AGAP006723-RA',
    cnv_region='2L:37,282,152-37,298,223',
    site_mask='arab',
    sample_query='sample_set == "1272-VO-GM-OPONDO-VMF00160" and taxon == "arabiensis" and sample_id!="VBS47740-6366STDY9854349" or country == "Tanzania" and taxon == "arabiensis" and unrestricted_use == True',
    color="country",
    linkage_method="complete",
    show=False
)
coeae2_7g_diplo.show(renderer="png")

coeae2_7g_diplo.write_image(os.path.join(figures_dir, "fig6_coeae2_7g_gambia_tanzania_diplptype_clustering.png"), scale=2)
```
